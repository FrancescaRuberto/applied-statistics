## Problem 1: Exploring Petal Length

### Introduction

The Iris dataset contains measurements of 150 flowers from three different species: Setosa, Versicolor and Virginica.

In this problem, I will focus on petal length and explore how it is distributed across all 150 flowers and within each species.

I will calculate the mean and standard deviation, create histograms and compare the data with normal distributions. The aim is to understand whether a normal distribution is a suitable model for petal length, both overall and for each individual species.

```python
# Import Iris Dataset
iris_dataset = pd.read_csv('/workspaces/applied-statistics/iris_dataset.csv')

# Print data
data = iris_dataset
print (data.head(10))
```

```python
# Let's have a look at the data we have inside
data.describe()
```

```python
# Let's build a plot for the petal length of the 150 flowers included in the dataset

x = data.petal_length
median_petal = np.median(x)

plt.figure(figsize = (10, 7))
plt.hist(x, bins = 20, color = "blue", edgecolor='black', alpha=0.6)

# axvline: axis vertical line
plt.axvline(median_petal, color="black", ls="--", label=f"Median: {median_petal:.2f} cm")
plt.legend()

plt.title("Petal Length")
plt.xlabel("Petal_Length")
plt.ylabel("Count")

plt.show()
```
